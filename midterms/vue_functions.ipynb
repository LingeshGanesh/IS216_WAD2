# **VUE functions**

## v-bind
v-bind lets an HTML attribute use a JavaScript/Vue value

<img :src="imageURL">

<a :href="website">Link</a>

<button :disabled="isDisabled">Submit</button>

<div :id="student.id">

<p :class="{ active: isActive }">

<p :style="{ color: textColor }">

combining it with class for buttons

<script setup>
import { ref } from 'vue'

const selected = ref(false)
</script>

<style>
.active {
    background-color: black;
    color: white;
}
</style>

<template>
    <input type="checkbox" v-model="selected">
    <!-- Apply 'active' class when selected = true -->
    <div :class="{ active: selected }">
        Hello
    </div>
</template>



## v-model
for full use with forms see [here](v_model.ipynb)

<!-- Text -->
<input v-model="name">

<!-- Textarea -->
<textarea v-model="description"></textarea>

<!-- Boolean checkbox -->
<input type="checkbox" v-model="agreed">

<!-- Multiple checkboxes → ref([]) -->
<input type="checkbox" value="Sports" v-model="hobbies">

<!-- Radio -->
<input type="radio" value="Male" v-model="gender">

<!-- Dropdown -->
<select v-model="course">
    <option value="IS">Information Systems</option>
</select>

## v-if => v-else-if => v-else

<!-- Basic -->
<p v-if="condition">Shown if true</p>

<!-- If / Else -->
<p v-if="condition">Yes</p>
<p v-else>No</p>

<!-- If / Else-if / Else -->
<p v-if="score >= 80">A</p>
<p v-else-if="score >= 70">B</p>
<p v-else>C</p>

for using to show and hide forms or other content

<script setup>
const show = ref(false)
</script>

<template>
    <button @click="show = !show">
        Toggle
    </button>

    <div v-if="show">
        Content
    </div>
</template>

## v-show
v-show is different from v-if in that the element still exists even when the condition is false. v-if removes it entirely from the page's DOM

<script setup>
import { ref } from 'vue'

const show = ref(false)
</script>

<template>
    <button @click="show = !show">
        Show / Hide
    </button>
    <div v-show="show" class="box">
        Hello!
    </div>
</template>

<style>
.box {
    background-color: lightblue;
    padding: 10px;
}
</style>

## v-for
repeats actions for a set number of items in array. For full applications see [here](v_for.ipynb)

<!-- Basic -->
<div v-for="item in items" :key="item.id">
    {{ item.name }}
</div>

<!-- With index -->
<div v-for="(item, index) in items" :key="item.id">
    {{ index + 1 }}. {{ item.name }}
</div>

<!-- Table -->
<tr v-for="student in students" :key="student.id">
    <td>{{ student.name }}</td>
    <td>{{ student.score }}</td>
</tr>

<!-- With action -->
<div v-for="student in students" :key="student.id">
    {{ student.name }}

    <button @click="deleteStudent(student.id)">
        Delete
    </button>
</div>

## Events

<!-- Click -->
<button @click="doSomething">
    Click
</button>

<!-- Pass parameter -->
<button @click="deleteStudent(student.id)">
    Delete
</button>

<!-- Form submission - prevent page refresh -->
<form @submit.prevent="submitForm">
    <button type="submit">Submit</button>
</form>

<!-- Run when value changes -->
<select @change="doSomething">
</select>

<!-- Run while input changes -->
<input @input="doSomething">

<!-- Enter key -->
<input @keyup.enter="submitForm">

## Computed
if we want our const to changed or be computed based on some other data or condition

<script setup>
    import { ref, computed } from 'vue'
</script>

<!--Use cases-->
<script>
// with if 
const grade = computed(() => {
    if (score.value >= 80) {
        return 'A'
    } else {
        return 'B'
    }
})

// with maths
const total = computed(() => {
    return price.value * quantity.value
})

// count items
const studentCount = computed(() => {
    return students.value.length
})

// filter an array
const honourStudents = computed(() => {
    return students.value.filter(
        student => student.gpa >= 4
    )
})

// count filtered items
const honourCount = computed(() => {
    return students.value.filter(
        student => student.gpa >= 4
    ).length
})

// combine strings
const firstName = ref('Hanafi')
const lastName = ref('Jaafar')

const fullName = computed(() => {
    return firstName.value + ' ' + lastName.value
})

// find one item
const selectedStudent = computed(() => {
    return students.value.find(
        student => student.id === 2
    )
})

// sort an array
const sortedStudents = computed(() => {
    return [...students.value].sort(
        (a, b) => b.gpa - a.gpa
    )
})
</script>
