## 1. Vue `ref()`

In `<script setup>`, use `.value` to access or change a ref.

In `<template>`, Vue automatically unwraps refs.

**Rule**
- Script → use `.value`
- Template → usually no `.value`


````javascript
const name = ref('')

name.value = 'John'
console.log(name.value)
````

Template example:

```html
<p>{{ name }}</p>
```


## 2. Array of Objects

Arrays can contain objects returned from an API.


````javascript
const stations = [
    { id: 1, name: 'Central Park' },
    { id: 2, name: 'Downtown' }
]

stations[0]        // whole first object
stations[0].name   // 'Central Park'
````

## 3. `push()`

Adds item(s) to the **end** of an array.


````javascript
let arr = [1, 2, 3]

arr.push(4)

console.log(arr)
// [1, 2, 3, 4]
````

With a Vue ref array:

```js
stations.value.push(newStation)
```


## 4. `pop()`

Removes the **last item**.


````javascript
let arr = [1, 2, 3]

const removed = arr.pop()

console.log(arr)
// [1, 2]

console.log(removed)
// 3
````

## 5. `unshift()`

Adds item(s) to the **front**.


````javascript
let arr = [2, 3]

arr.unshift(1)

console.log(arr)
// [1, 2, 3]
````

## 6. `shift()`

Removes the **first item**.


````javascript
let arr = [1, 2, 3]

arr.shift()

console.log(arr)
// [2, 3]
````

## 7. `splice()`

General syntax:

```js
array.splice(startIndex, deleteCount, itemsToAdd)
```

Common uses:
- Remove one item → `arr.splice(index, 1)`
- Remove multiple → `arr.splice(index, numberToDelete)`
- Add without removing → `arr.splice(index, 0, newItem)`
- Replace → `arr.splice(index, 1, replacement)`


````javascript
let arr = ['A', 'B', 'C', 'D']

arr.splice(1, 2)

console.log(arr)
// ['A', 'D']
````

````javascript
let arr2 = ['A', 'C']

arr2.splice(1, 0, 'B')

console.log(arr2)
// ['A', 'B', 'C']
````

````javascript
let arr3 = ['A', 'B', 'C']

arr3.splice(1, 1, 'X')

console.log(arr3)
// ['A', 'X', 'C']
````

## 8. `find()`

Returns the **first matching item**.

Syntax:

```js
array.find(item => condition)
```

If the array contains objects, `find()` returns the **entire matching object**.


````javascript
const stations = [
    { id: 1, name: 'Central Park' },
    { id: 2, name: 'Downtown' }
]

const target = stations.find(
    station => station.name === 'Downtown'
)

console.log(target)
// { id: 2, name: 'Downtown' }

console.log(target.id)
// 2
````

If nothing matches, `find()` returns:

```js
undefined
```

So it is safer to check:

```js
if (target) {
    console.log(target.id)
}
```


## 9. `find()` with Vue `ref([])`

If the array is stored inside a Vue ref, use `.value` in the script.


````javascript
const target = stations.value.find(
    station => station.name === selectedStation.value
)
````

## 10. `findIndex()`

Returns the **index** of the matching item instead of the item itself.


````javascript
const index = stations.findIndex(
    station => station.name === 'Downtown'
)

console.log(index)
// 1
````

Useful with `splice()`:


````javascript
const index = stations.findIndex(
    station => station.id === 2
)

if (index !== -1) {
    stations.splice(index, 1)
}
````

## 11. `find()` vs `findIndex()`

```js
stations.find(s => s.name === 'Downtown')
```

Returns the object:

```js
{ id: 2, name: 'Downtown' }
```

```js
stations.findIndex(s => s.name === 'Downtown')
```

Returns the index:

```js
1
```

**Remember**
- `find()` → give me the item
- `findIndex()` → tell me where it is


## 12. Arrow Function Syntax

Short form:


````javascript
station => station.name === 'Downtown'
````

Equivalent long form:


````javascript
function(station) {
    return station.name === 'Downtown'
}
````

With curly braces, remember to use `return`:


````javascript
station => {
    return station.name === 'Downtown'
}
````

## 15. Axios Error Handling

Axios HTTP status codes are usually available at:

```js
error.response.status
```


````javascript
try {
    await axios.put(connectionAPI, toSend)
} catch (error) {
    console.log(error)

    if (error.response?.status === 400) {
        alert('Bad request')
    }
}
````

Useful values:

```js
error.response.status
error.response.data
```

- `400` → Bad request
- `404` → Not found
- `500` → Backend/server error


## 17. Vue `<select>` with `v-model`

Use `:` before `value` when binding a JavaScript expression.

```html
<select v-model="selectedStation">
    <option value="">Select a station</option>

    <option
        v-for="station in stations"
        :value="station.name"
    >
        {{ station.name }}
    </option>
</select>
```

Difference:

```html
value="station.name"
```

means the literal text `"station.name"`.

```html
:value="station.name"
```

uses the actual value stored in `station.name`.


## 18. Resetting a Vue `<select>`

If the default option is:

```html
<option value="">Select a station</option>
```

reset the ref with:

```js
selectedStation.value = ''
```

Because `selectedStation` is a Vue ref.
