# **Modularization**

## controllers
The controller folder contains files with most of the javascript processing code which ends with the rendering

## data

The data folder contains the stored data, usually in JSON files

## models

The models folder contains the code that is responsible for data input and preparation.

## routes

The routes folders contain the actions that are to be performed when each route is accessed. These usually include just the get commands

## views

The views folder contains the html and webpage front end that the user sees when they visit the webpage. They are usually ejs files
