# **Form Processing using ejs**

```python
const express = require('express');
const server = express();

const hostname = 'localhost';
const port = 8000;

server.set("view engine", "ejs");
```

when using ejs, you'll often need to use 2 get/post commands rather than just one. This is because one needs to be used to get the ejs template and the second will be used to get the command which needs to be rendered.

```python
server.get("/", (req, res) => {
  res.render("like", { message: null, bookmarked: false });
});
```

When rendering your ejs template, you need to put in the variables that you want to use in the render command in {} or they will load in as undefined

```python
server.get("/process-like", (req, res) => {
  const feeling = req.query.feeling;
  const bookmarked = req.query.bookmark ? true : false;

  res.render("like", { message, bookmarked });
});
```
