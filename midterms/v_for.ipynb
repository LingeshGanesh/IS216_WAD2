# **Applications for v-for**

## simple list

<script setup>
import { ref } from 'vue'

const fruits = ref(['Apple', 'Banana', 'Orange'])
</script>

<template>
    <ul>
        <li v-for="fruit in fruits">
            {{ fruit }}
        </li>
    </ul>
</template>

## Array of objects

<div v-for="student in students" :key="student.id">
    <p>{{ student.name }}</p>
    <p>{{ student.score }}</p>
</div>

## Tables

<table>
    <thead>

## Dropdown

<script setup>
    const courses = ref([
    { id: 1, name: 'IS216' },
    { id: 2, name: 'IS212' },
    { id: 3, name: 'IS210' }
])

const selectedCourse = ref('')
</script>

<select v-model="selectedCourse">
    <option
        v-for="course in courses"
        :key="course.id"
        :value="course.id"
    >
        {{ course.name }}
    </option>
</select>

## Returning values only if they meet certain conditions

<div v-for="student in students" :key="student.id">
    {{ student.name }}
    <span v-if="student.score > 90">
        ⭐
    </span>
</div>

<!--Returning list based on conditon-->
<ul>
    <template v-for="student in students">
        <li v-if="data.field>condition">{{data.info}}</li>
    </template>
</ul>

## With buttons
