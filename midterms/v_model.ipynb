# **v-model with forms**

## Text inputs

<script setup>
import { ref } from 'vue'

const name = ref('')
</script>

<template>
    <input type="text" v-model="name">
    <p>{{ name }}</p>
</template>

## Textarea

<script setup>
import { ref } from 'vue'

const description = ref('')
</script>

<template>
    <textarea v-model="description"></textarea>
    <p>{{ description }}</p>
</template>

## Checkbox (Boolean)

<script setup>
import { ref } from 'vue'

const agreed = ref(false)
</script>

<template>
    <input type="checkbox" v-model="agreed">

    <p>{{ agreed }}</p>
</template>

## Checkbox (Multiple/Array)

<script setup>
import { ref } from 'vue'

const hobbies = ref([])
</script>

<template>
    <input type="checkbox" value="Sports" v-model="hobbies"> Sports
    <input type="checkbox" value="Music" v-model="hobbies"> Music
    <input type="checkbox" value="Gaming" v-model="hobbies"> Gaming

    <p>{{ hobbies }}</p>
</template>
