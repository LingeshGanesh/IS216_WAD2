## Standard import statement
import { ref, computed, onMounted } from 'vue'

## Get data from api on mount

**When referencing values that have been pulled from the GET, use "data.value" inside <script setup> and use it as {{data}} inside template**

<script setup>
    import { ref, onMounted } from 'vue'
    import axios from 'axios'

    const dataFrame = ref([])

    const getData = async () => {
    const response = await axios.get(
        'http://localhost:8000/apiLINK'
    )

    dataFrame.value = response.data
    }

    onMounted(() => {
    getData()
    })
    
</script>

onMounted(getData()) also works to reduce redundancy

<!--Cycling through items in a list of data-->
<div v-for="student in students">
    {{ student.name }}
</div>

## Post data to API

await axios.post(URL, DATA)

<script setup>
    import { ref } from 'vue'
    import axios from 'axios'
    
    const name = ref('')
    const gpa = ref('')

    const addStudent = async () => {

        const newStudent = {
            name: name.value,
            gpa: gpa.value
        }

        await axios.post(
            'http://localhost:8000/students',
            newStudent
        )
    }
</script>

<template>
    <input v-model="name">
    <input v-model="gpa">
    <button @click="addStudent">Add</button>
</template>

## Put (Update) Data in API

<script setup>
import { ref } from 'vue'
import axios from 'axios'

const id = ref('')
const name = ref('')
const gpa = ref('')

const updateStudent = async () => {

    const updatedStudent = {
        name: name.value,
        gpa: gpa.value
    }

    await axios.put(
        `http://localhost:8000/students/${id.value}`,
        updatedStudent
    )
}
</script>

<template>
    <input v-model="id">
    <input v-model="name">
    <input v-model="gpa">
    <button @click="updateStudent">
        Update
    </button>
</template>

## Delete data from API

<script setup>
    import { ref } from 'vue'
    import axios from 'axios'

    const id = ref('')

    const deleteStudent = async () => {
        try {
            await axios.delete(`http://localhost:8000/students/${id.value}`)
        } catch (error) {
            console.log(error)
    }}
</script>

<template>
    <input v-model="id">
    <button @click="deleteStudent">
        Delete
    </button>
</template>
