# **Disabling forms based on conditions**

<script setup>
import { ref } from 'vue'

const name = ref('')
const course = ref('')
const gender = ref('')
const agreed = ref(false)
const status = ref('pending')
</script>

<template>

    <!-- ======================================== -->
    <!-- DISABLE SYNTAX                           -->
    <!-- ======================================== -->

    <!-- Always disabled -->
    <input disabled>

    <!-- Dynamically disabled -->
    <!-- condition TRUE  = disabled -->
    <!-- condition FALSE = enabled -->
    <input :disabled="condition">


    <!-- ======================================== -->
    <!-- TEXT / NUMBER INPUT                      -->
    <!-- ======================================== -->

    <!-- Disable until name is entered -->
    <input
        type="text"
        v-model="name"
        :disabled="course === ''"
    >

    <!-- Works the same for number -->
    <input
        type="number"
        v-model="age"
        :disabled="condition"
    >


    <!-- ======================================== -->
    <!-- TEXTAREA                                 -->
    <!-- ======================================== -->

    <textarea
        v-model="description"
        :disabled="condition"
    ></textarea>


    <!-- ======================================== -->
    <!-- DROPDOWN / SELECT                        -->
    <!-- ======================================== -->

    <!-- Disable entire dropdown -->
    <select
        v-model="course"
        :disabled="name === ''"
    >
        <option value="">Select Course</option>
        <option value="IS">IS</option>
        <option value="CS">CS</option>
    </select>


    <!-- Disable individual option -->
    <select v-model="course">

        <option value="">Select Course</option>

        <option
            value="IS"
            :disabled="condition"
        >
            IS
        </option>

        <option value="CS">
            CS
        </option>

    </select>


    <!-- ======================================== -->
    <!-- CHECKBOX                                 -->
    <!-- ======================================== -->

    <!-- Disable checkbox -->
    <input
        type="checkbox"
        v-model="agreed"
        :disabled="condition"
    >
    I agree


    <!-- ======================================== -->
    <!-- RADIO BUTTONS                            -->
    <!-- ======================================== -->

    <!-- Disable individual radio option -->
    <input
        type="radio"
        value="Male"
        v-model="gender"
        :disabled="condition"
    >
    Male

    <input
        type="radio"
        value="Female"
        v-model="gender"
        :disabled="condition"
    >
    Female


    <!-- ======================================== -->
    <!-- BUTTON                                   -->
    <!-- ======================================== -->

    <!-- Disable until name entered -->
    <button :disabled="name === ''">
        Submit
    </button>

    <!-- Disable until checkbox checked -->
    <button :disabled="!agreed">
        Submit
    </button>

    <!-- Disable until radio selected -->
    <button :disabled="gender === ''">
        Submit
    </button>

    <!-- Disable until dropdown selected -->
    <button :disabled="course === ''">
        Submit
    </button>


    <!-- ======================================== -->
    <!-- MULTIPLE CONDITIONS                      -->
    <!-- ======================================== -->

    <!--
        Disable if ANY requirement is missing
        Enabled only when ALL are completed
    -->
    <button
        :disabled="
            name === '' ||
            course === '' ||
            gender === '' ||
            !agreed
        "
    >
        Submit
    </button>


    <!-- ======================================== -->
    <!-- DISABLE BASED ON STATUS / VALUE          -->
    <!-- ======================================== -->

    <!-- Disable Approve when already approved -->
    <button :disabled="status === 'approved'">
        Approve
    </button>

    <!-- Disable Deny when already denied -->
    <button :disabled="status === 'denied'">
        Deny
    </button>


    <!-- ======================================== -->
    <!-- INSIDE V-FOR                             -->
    <!-- ======================================== -->

    <tr
        v-for="request in requests"
        :key="request.id"
    >
        <td>{{ request.name }}</td>

        <td>
            <!-- Each button checks THAT request's status -->

            <button
                :disabled="request.status === 'approved'"
            >
                Approve
            </button>

            <button
                :disabled="request.status === 'denied'"
            >
                Deny
            </button>
        </td>
    </tr>

</template>
